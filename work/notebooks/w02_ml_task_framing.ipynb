## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My Lane as an ML Task (Type)

I selected the **Content Refresh Prioritization** lane. This is primarily a **ranking** problem because the goal is to rank web pages based on their likelihood of needing a content refresh. Instead of simply classifying pages as good or bad, the system prioritizes which pages should be reviewed first. Ranking helps SEO teams focus on the pages that are most likely to benefit from updates.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**## Target or Proxy**

The target is whether a page is experiencing a declining trend. Since the dataset already contains the `trend_direction` column, I use the following proxy label:

`is_declining_label = (trend_direction == "down")`

This label is derived from an observed outcome in the dataset and serves as the prediction target for prioritizing content review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*



The primary evaluation metric is **Precision@50**.

This measures how many of the top 50 pages recommended by the model are actually declining. A higher Precision@50 means the ranking system is better at identifying the pages that deserve attention first, making it more useful for SEO analysts.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (
    df["trend_direction"].str.lower() == "down"
).astype(int)

lane_df = df[
    [
        "content_type",
        "impressions_90d",
        "ctr",
        "word_count",
        "content_age_days",
        "trend_direction",
        "is_declining_label"
    ]
]

print("Unit of Analysis: One row = One Web Page")
print("Dataset Shape:", lane_df.shape)

lane_df.head()
```

## The Unit of Analysis

The unit of analysis is one web page.

Each row in the dataframe represents a single web page with measurable features such as content type, impressions, click-through rate (CTR), word count, content age, trend direction, and the generated decline label. These features help determine whether a page should be prioritized for content refresh.

## 5. Why ML beats a fixed rule here


A fixed rule can only consider a small number of conditions, such as content age or impressions. However, page performance depends on several interacting factors including CTR, impressions, content age, and content type.

Machine learning can learn these relationships automatically and produce a better ranking than a simple hand-written rule. This enables more effective prioritization of pages for review and supports data-driven decision making.
